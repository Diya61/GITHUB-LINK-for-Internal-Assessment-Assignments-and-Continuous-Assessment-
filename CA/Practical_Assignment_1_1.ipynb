```python

class VacuumEnvironment:

    def __init__(self, room_a, room_b):
        self.rooms = {
            "A": room_a,
            "B": room_b
        }


class SimpleReflexVacuumAgent:

    def __init__(self, location):
        self.location = location
        self.actions = 0
        self.suck_actions = 0
        self.move_actions = 0

    def perceive_and_act(self, environment):

        status = environment.rooms[self.location]

        if status == "Dirty":
            print(f"Location: Room {self.location} | Status: Dirty | Action: SUCK")

            environment.rooms[self.location] = "Clean"
            self.suck_actions += 1
            self.actions += 1

        elif self.location == "A":
            print("Location: Room A | Status: Clean | Action: MOVE RIGHT")

            self.location = "B"
            self.move_actions += 1
            self.actions += 1

        else:
            print("Location: Room B | Status: Clean | Action: MOVE LEFT")

            self.location = "A"
            self.move_actions += 1
            self.actions += 1

    def clean_environment(self, environment):

        while "Dirty" in environment.rooms.values():
            self.perceive_and_act(environment)

        print(f"Location: Room {self.location} | Status: Clean | Action: STOP")

        return self.actions


# Create environment
environment = VacuumEnvironment("Dirty", "Dirty")

# Create agent
agent = SimpleReflexVacuumAgent("A")

print("Initial Environment:")
print(environment.rooms)
print()

# Clean environment
agent.clean_environment(environment)

print()
print("Final Environment:")
print(environment.rooms)

print()
print("Performance Analysis:")
print("Total Actions:", agent.actions)
print("Suck Actions:", agent.suck_actions)
print("Movement Actions:", agent.move_actions)
```

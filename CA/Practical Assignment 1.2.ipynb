```python

class KnowledgeBase:
    def __init__(self):
        self.facts = []
        self.rules = []

    def add_fact(self, fact):
        self.facts.append(fact)

    def add_rule(self, condition, conclusion):
        self.rules.append((condition, conclusion))

    def infer(self):
        for condition, conclusion in self.rules:
            if condition(self.facts):
                if conclusion not in self.facts:
                    self.facts.append(conclusion)
                    print("Inferred Fact:", conclusion)

    def display_facts(self):
        print("\nKnowledge Base Facts:")
        for fact in self.facts:
            print("-", fact)


kb = KnowledgeBase()

# User Input
print("Enter symptoms: fever, cough, headache")
symptoms = input(
    "Enter symptoms separated by commas: "
).lower().split(",")

for symptom in symptoms:
    symptom = symptom.strip()
    if symptom:
        kb.add_fact(symptom)

# Rules
kb.add_rule(
    lambda facts: "fever" in facts and "cough" in facts,
    "possible_respiratory_infection"
)

kb.add_rule(
    lambda facts: "fever" in facts,
    "monitor_body_temperature"
)

kb.add_rule(
    lambda facts: "headache" in facts and "fever" in facts,
    "seek_medical_advice"
)

# Apply Rules
print("\nApplying Rules:")
kb.infer()

# Display Results
kb.display_facts()
```
